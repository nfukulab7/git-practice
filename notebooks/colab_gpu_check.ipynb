# Colab GPU environment check

Before running this notebook, choose **Runtime > Change runtime type** and select a GPU only if the workload needs one. This notebook records the runtime environment without assuming a specific GPU model.

```python
import os
import platform
import shutil
import subprocess
import sys

def command_output(*command: str) -> str:
    completed = subprocess.run(command, capture_output=True, text=True, check=False)
    return (completed.stdout or completed.stderr).strip()

print(f"Python: {sys.version.split()[0]}")
print(f"Platform: {platform.platform()}")
print(f"Working directory: {os.getcwd()}")

if shutil.which("nvidia-smi"):
    print("GPU detected by nvidia-smi:")
    print(command_output("nvidia-smi", "--query-gpu=name,driver_version,memory.total", "--format=csv,noheader"))
else:
    print("No NVIDIA GPU detected. Select a GPU runtime only when needed.")
```

```python
try:
    import torch
except ImportError:
    print("PyTorch is not installed in this runtime.")
else:
    print(f"PyTorch: {torch.__version__}")
    print(f"CUDA available: {torch.cuda.is_available()}")
    if torch.cuda.is_available():
        print(f"CUDA runtime: {torch.version.cuda}")
        print(f"Device: {torch.cuda.get_device_name(0)}")
        left = torch.tensor([1.0, 2.0], device="cuda")
        right = torch.tensor([3.0, 4.0], device="cuda")
        print(f"GPU smoke test: {(left + right).tolist()}")
```

```python
if shutil.which("git") and os.path.isdir(".git"):
    print(f"Branch: {command_output('git', 'branch', '--show-current')}")
    print(f"Commit: {command_output('git', 'rev-parse', 'HEAD')}")
else:
    print("No Git checkout detected. Open from GitHub or clone the repository to record a commit.")
```
